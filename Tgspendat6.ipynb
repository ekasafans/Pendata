##Penjelasan Gambar Di Atas

1. Input Data (Excel Reader)

Tahap pertama adalah membaca data dari file Excel menggunakan node Excel Reader.
Data ini berisi atribut (fitur) dan satu kolom target (label) yang akan diprediksi.

2. Pembagian data (tabel partitioner)
Node Table Partitioner digunakan untuk membagi data menjadi:

Data training (untuk melatih model)
Data testing (untuk menguji model)

Biasanya pembagian dilakukan dengan perbandingan, misalnya:

67% training
33% testing

Metode yang digunakan adalah Stratified Sampling, yaitu pembagian data berdasarkan proporsi kelas agar seimbang.

3. Pewarnaan Data (Color Manager & Color Appender)
Color Manager: memberi warna pada data training berdasarkan kelas (label)
Color Appender: menambahkan warna pada data testing agar mudah divisualisasikan

Tahap ini membantu dalam visualisasi hasil klasifikasi.

4. Pembuatan Model (Decision Tree Learner)
Node Decision Tree Learner digunakan untuk membangun model pohon keputusan.

Hasilnya terlihat pada gambar tengah berupa struktur pohon:

Setiap node menunjukkan kondisi (aturan)
Setiap cabang menunjukkan hasil keputusan berdasarkan kondisi tertentu
Setiap daun (leaf) menunjukkan hasil klasifikasi (Yes / No)

Contoh pada gambar:

Total data: 9
Kelas “Yes”: 66.7% (6 data)
Kelas “No”: 33.3% (3 data)

Kemudian pohon membagi data berdasarkan kondisi tertentu (misalnya atribut tertentu bernilai True/False).

5. Prediksi (Decision Tree Predictor)
Node ini digunakan untuk:

Menerapkan model decision tree ke data testing
Menghasilkan prediksi kelas (Yes / No)

6. Evaluasi Model (Scorer)
Node Scorer digunakan untuk mengevaluasi performa model, seperti:

Akurasi
Confusion matrix
Perbandingan hasil prediksi dengan data asli

Penjelasan hasil decision tree
Node awal (root):
66.7% = Yes
33.3% = No
Cabang kiri:
Didominasi oleh kelas Yes (80%)
Cabang kanan:
Data terbagi seimbang antara Yes dan No (50% : 50%)
