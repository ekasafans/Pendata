##2. Grafik Scatter Plot
Titik-titik biru pada grafik disebut scatter plot atau diagram pencar. Scatter plot digunakan untuk melihat pola hubungan antara data X dan Y.

Dari grafik terlihat bahwa:

semakin besar nilai X,
nilai Y juga cenderung meningkat.

Hal ini menunjukkan adanya hubungan linear positif.

##3. Garis Regresi Linier
GeoGebra menghasilkan garis regresi:

y=1.0769230769231x


image.png

##4. Arti Persamaan Regresi
Persamaan:

y=1.0769x

memiliki arti:

y = variabel dependen
x = variabel independen
1.0769 = koefisien regresi/slope

Koefisien tersebut menunjukkan bahwa:

setiap kenaikan 1 nilai pada x,
maka nilai y akan meningkat sekitar 1.0769.
